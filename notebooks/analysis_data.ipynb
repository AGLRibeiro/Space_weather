```python
# Expandir a lista de instrumentos em linhas
df_temp = data[["activityID", "instruments"]].explode(
    "instruments"
)

# Remover valores nulos
#df_temp = df_temp.dropna(subset=["instruments"])

# Converter os dicionários em colunas
df_instruments = pd.json_normalize(
    df_temp["instruments"]
)

# Recuperar a chave de relacionamento
df_instruments["activityID"] = (
    df_temp["activityID"].values
)

# Reorganizar as colunas
df_instruments = df_instruments[
    ["activityID", "displayName"]
]

print(df_instruments)
```

```python
# Expandir a lista de instrumentos em linhas
df_temp = data[["activityID", "instruments"]].explode(
    "instruments"
)

# Remover valores nulos
df_temp = df_temp.dropna(subset=["instruments"])

# Converter os dicionários em colunas
df_instruments = pd.json_normalize(
    df_temp["instruments"]
)

# Recuperar a chave de relacionamento
df_instruments["activityID"] = (
    df_temp["activityID"].values
)

# Reorganizar as colunas
df_instruments = df_instruments[
    ["activityID", "displayName"]
]

print(df_instruments)
```
