```python
import os
import pandas as pd
from dotenv import load_dotenv
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().resolve().parent

DATA_DIR = BASE_DIR / "data" / "bronze"

data = pd.read_json(DATA_DIR / "CMEAnalysis_data.json")
```

```python
def explode_and_normalize(df = pd.DataFrame) -> pd.DataFrame:
    df = df.explode()
    df = pd.json_normalize(df)
    return df
```
